```python
import os
import json
import numpy as np
import matplotlib.pyplot as plt

# =========================
# 全局画图设置
# =========================
plt.rcParams['font.size'] = 16
plt.rcParams['font.sans-serif'] = ['Arial', 'SimHei']
plt.rcParams['axes.unicode_minus'] = False

# =========================
# 1. 定义实验文件夹
# =========================
experiment_dirs = [
    r"data\compare_data\250909 v3 same weights 55520 0.99 seed 30",
    r"data\compare_data\250909 v3 same weights 55520 0.99 seed 42",
    r"data\compare_data\250909 v3 same weights 55520 0.99 seed 100",
    # r"data\compare_data\260421_multi_weight_seed61",
    # r"data\compare_data\260421_multi_weight_seed81",
    # r"data\compare_data\260421_multi_weight_seed31",
    
]

# =========================
# 2. 读取单个实验函数
# =========================
def read_progress_json(exp_dir, reward_key="rollout/ep_rew_mean", episode_key="time/episodes"):
    reward_list = []
    episode_list = []

    progress_path = os.path.join(exp_dir, "progress.json")
    if not os.path.exists(progress_path):
        raise FileNotFoundError(f"未找到文件: {progress_path}")

    with open(progress_path, "r", encoding="utf-8") as f:
        for line in f:
            if line.strip():
                data = json.loads(line)
                if reward_key in data and episode_key in data:
                    reward_list.append(data[reward_key])
                    episode_list.append(data[episode_key])

    return np.array(episode_list), np.array(reward_list)

# =========================
# 3. 读取所有实验
# =========================
all_episodes = []
all_rewards = []

for exp_dir in experiment_dirs:
    episodes, rewards = read_progress_json(exp_dir)
    all_episodes.append(episodes)
    all_rewards.append(rewards)

# =========================
# 4. 对齐长度
#    如果长度不同，裁剪到最短长度
# =========================
min_len = min(len(x) for x in all_rewards)

aligned_episodes = all_episodes[0][:min_len]   # 默认所有实验记录频率相同
aligned_rewards = np.array([x[:min_len] for x in all_rewards])

# shape: (n_runs, min_len)
# aligned_rewards[i, t] 表示第 i 个实验在第 t 个记录点的 reward

# =========================
# 5. 计算均值和标准差
# =========================
mean_rewards = np.mean(aligned_rewards, axis=0)
std_rewards = np.std(aligned_rewards, axis=0, ddof=1)  # 样本标准差

# =========================
# 6. 绘图
# =========================
plt.figure(figsize=(10, 6))

# # 可选：先画每个 seed 的淡线
# for i in range(aligned_rewards.shape[0]):
#     plt.plot(
#         aligned_episodes,
#         aligned_rewards[i],
#         color="gray",
#         alpha=0.35,
#         linewidth=1
#     )
    
line_colors = ["#d58abb", "#d28639", "#ba463a"]

label_seeds = ["1", "2", "3"]

for i in range(aligned_rewards.shape[0]):
    seed_name = label_seeds[i]
    label = f"Seed {seed_name} (Main text)" if seed_name == "2" else f"Seed {seed_name}"
    plt.plot(
        aligned_episodes,
        aligned_rewards[i],
        color=line_colors[i % len(line_colors)],
        alpha=0.6,
        linewidth=1.2,  
        label=label
    )

# 均值曲线
plt.plot(
    aligned_episodes,
    mean_rewards,
    color="#21908c",
    linewidth=2.8,
    label="Mean reward across seeds"
)

# 标准差阴影
plt.fill_between(
    aligned_episodes,
    mean_rewards - 1 * std_rewards,
    mean_rewards +  1 * std_rewards,
    color="#21908c",
    alpha=0.2,
    label="±1 SD"
)

plt.xlabel(" Episodes")
plt.ylabel("Average Reward per Episode")

# 图例加粗避免看不清颜色
legend = plt.legend(loc="lower right")
for h in legend.legendHandles:
    if hasattr(h, "set_linewidth"):
        h.set_linewidth(3.0)
    # if hasattr(h, "set_alpha"):
    #     h.set_alpha(1.0)
plt.grid(False)
plt.tight_layout()

# plt.savefig("data/plot_articles/episode_reward_seed_comparison_new.png", dpi=600, bbox_inches='tight')
plt.savefig("data/plot_articles/episode_reward_seed_comparison_new.tif", dpi=600, bbox_inches='tight', pil_kwargs={"compression": "tiff_lzw"} )
plt.show()

```
