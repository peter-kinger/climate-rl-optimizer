```python
import numpy as np
import matplotlib.pyplot as plt

# Define temperature range
T_a = np.linspace(0, 3, 300)

# Parameters for piecewise function
T_thresh = 1.5     # threshold for piecewise
R_thresh = -100    # reward at the threshold
alpha = 3.0        # shape parameter for the exponential part
linear_rate = 100  # additional penalty per unit above threshold

# Compute reward: exponential approach to R_thresh for Ta <= T_thresh,
# linear decrease beyond.
reward = np.where(
    T_a <= T_thresh,
    R_thresh * (1 - np.exp(-alpha * T_a)) / (1 - np.exp(-alpha * T_thresh)),
    R_thresh - linear_rate * (T_a - T_thresh)
)

# Plotting
plt.figure(figsize=(6, 4))
plt.plot(T_a, reward)
plt.xlabel(r'$T_a$ (°C)')
plt.ylabel('Reward')
plt.title('Piecewise Exponential then Linear Reward Function')
plt.grid(True)
plt.show()
```

```python

import numpy as np
import matplotlib.pyplot as plt

# 参数设置
kappa = 3.0          # 控制指数陡峭度
T_thresh = 1.5       # 拐点温度
R_thresh = -100.0    # 在 Ta = 1.5 时的 reward

# 在 [0, 1.5] 范围内离散 Ta
T_a = np.linspace(0, T_thresh, 300)

# 纯指数曲线：保证 r(0)=0, r(1.5)=-100
reward = R_thresh * (np.exp(kappa * T_a) - 1) / (np.exp(kappa * T_thresh) - 1)

# 绘图
plt.figure(figsize=(6,4))
plt.plot(T_a, reward, label=f'κ={kappa}')
plt.axhline(0, color='gray', linewidth=0.5)
plt.axvline(T_thresh, color='gray', linestyle='--', linewidth=0.5)
plt.scatter([0, T_thresh], [0, R_thresh], color='red')
plt.text(0, 0, ' (0,0)', va='bottom', ha='right')
plt.text(T_thresh, R_thresh, f' ({T_thresh},{R_thresh})', va='top', ha='left')
plt.xlabel(r'$T_a$ (°C)')
plt.ylabel('Reward')
plt.title('Pure Exponential Reward Curve')
plt.grid(True)
plt.legend()
plt.show()
```

```python

import numpy as np
import matplotlib.pyplot as plt

# 参数设置
T0 = 1.064859395    # 起始点 Ta，使 reward=0
T1 = 1.5            # 终点 Ta，使 reward=-100
R1 = -100.0         # 在 Ta=T1 时的 reward
kappa = 3.0         # 控制指数陡峭度

# 生成温度序列
T_a = np.linspace(0, 3, 300)

# 计算 reward
reward = np.zeros_like(T_a)
mask = (T_a >= T0) & (T_a <= T1)
reward[mask] = R1 * (np.exp(kappa * (T_a[mask] - T0)) - 1) \
               / (np.exp(kappa * (T1 - T0)) - 1)

# 绘图
plt.figure(figsize=(6,4))
plt.plot(T_a, reward, label='指数段')
plt.axhline(0, color='gray', linewidth=0.5)
plt.axvline(T0, color='gray', linestyle='--', linewidth=0.5)
plt.axvline(T1, color='gray', linestyle='--', linewidth=0.5)
plt.scatter([T0, T1], [0, R1], color='red')
plt.text(T0, 0, f'  ({T0:.3f}, 0)', va='bottom', ha='left')
plt.text(T1, R1, f'  ({T1}, {R1})', va='top', ha='left')
plt.xlabel(r'$T_a$ (°C)')
plt.ylabel('Reward')
plt.title('指数曲线：从 $r=0$ 到 $r=-100$')
plt.grid(True)
plt.legend()
plt.show()
```

```python

import numpy as np
import matplotlib.pyplot as plt

# 参数设置
kappa    = 3.0              # 控制指数陡峭度
T0       = 1.064859395      # 起始点 Ta，使 reward(T0)=0
T_thresh = 1.5              # 终点 Ta，使 reward(T_thresh)=-100
R_thresh = -100.0           # 在 Ta = T_thresh 时的 reward

# 生成温度序列
T_a = np.linspace(0, 3, 300)

# 先全部置 0
reward = np.zeros_like(T_a)

# 仅在 [T0, T_thresh] 区间内用指数曲线
reward = np.where(
    T_a <  T0,       0,
    np.where(
        T_a <= T_thresh,
        R_thresh * (np.exp(kappa*(T_a - T0)) - 1)
                 / (np.exp(kappa*(T_thresh - T0)) - 1),
        R_thresh      # T_a > T_thresh 时保持 -100
    )
)


# 绘图
plt.figure(figsize=(6,4))
plt.plot(T_a, reward, label='指数曲线段')
plt.axhline(0, color='gray', linewidth=0.5)
plt.axvline(T0, color='gray', linestyle='--', linewidth=0.5)
plt.axvline(T_thresh, color='gray', linestyle='--', linewidth=0.5)
plt.scatter([T0, T_thresh], [0, R_thresh], color='red')
plt.text(T0, 0,   f' ({T0:.3f},  0)',  va='bottom', ha='left')
plt.text(T_thresh, R_thresh, f' ({T_thresh}, {R_thresh})',
         va='top', ha='left')
plt.xlabel(r'$T_a$ (°C)')
plt.ylabel('Reward')
plt.title('分段指数曲线：在 $T_0$ 到 $T_{\\rm th}$ 之间从 0 下降到 -100')
plt.grid(True)
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# ——— 参数设置 ———
T1, R1 = 1.5, -10      # 直线－指数分界点：Ta=1.5 时 reward=-10
T2, R2 = 1.76, -100    # 指数段终点：Ta=1.76 时 reward=-100
kappa   = 10.0         # 控制指数曲线陡峭度，可根据需要调整

# ——— 温度序列 ———
T_a = np.linspace(0, 3, 300)

# ——— 计算 reward ———
reward = np.empty_like(T_a)

# 前半段：线性
mask_lin = T_a <= T1
reward[mask_lin] = (R1 / T1) * T_a[mask_lin]

# 后半段：指数
mask_exp = T_a > T1
# 归一化常数 C，使得 exp 部分在 T2 时恰好从 R1 跳到 R2
C = (R2 - R1) / (np.exp(kappa * (T2 - T1)) - 1)
reward[mask_exp] = R1 + C * (np.exp(kappa * (T_a[mask_exp] - T1)) - 1)

# ——— 绘图 ———
plt.figure(figsize=(6,4))
plt.plot(T_a, reward, label='piecewise: linear→exponential')
plt.axvline(T1, color='gray', linestyle='--', label=f'$T_1={T1}$')
plt.axvline(T2, color='gray', linestyle=':', label=f'$T_2={T2}$')
plt.scatter([0, T1, T2], [0, R1, R2], color='red')
plt.text(0, 0,    ' (0,0)',    va='bottom', ha='right')
plt.text(T1, R1,  f' ({T1},{R1})',  va='bottom', ha='left')
plt.text(T2, R2,  f' ({T2},{R2})',  va='top',    ha='left')
plt.xlabel(r'$T_a$ (°C)')
plt.ylabel('Reward')
plt.title('Piecewise Linear then Exponential Reward')
plt.grid(True)
plt.legend()
plt.show()
```
