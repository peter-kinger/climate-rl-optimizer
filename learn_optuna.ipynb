如果你有也这个问题，看这篇就对了！虽然你可能听过 Sklearn 的 GridSearchCV 网格搜索同样也是暴力的找出最佳参数，或是使用 RandomizedSearchCV 随机搜索指定超参数的范围并随机的抽取参数进⾏训练，其它们的共同缺点是非常耗时与占用机器资源。

实例学习：调节 x,y 值知道如何取得最小值，

```python
import optuna 
def objective(trial):
    x = trial.suggest_float("x", -7, 7) # x 即是想要调节的超参数对象，其实等同于 grid search 中的超参数的 {"x": np.arrange(-7, 7)}
    y = trial.suggest_float("y", -7, 7)
    return (x - 1) ** 2 + (y + 3) ** 2

study = optuna.create_study(direction="minimize") # 目标函数是最小化
study.optimize(objective, n_trials=1000) # 进行100次试验, 实验次数越多，结果越接近于全局最优解
study.best_params # 输出最优参数

```
